```python
# env: routesim

import json
import pandas as pd
import rxnutils

from rxnutils import routes
from rxnutils.routes import readers
from rxnutils.routes.readers import read_reactions_dataframe
from rxnutils.routes import comparison
from rxnutils.routes.comparison import simple_route_similarity

import os
```

```python
DIR_IN = '../data/output/predicted_routes/mapped_rxns/'

FNAME_IN_ALL =[]

FNAME_OUT = '../data/output/similarity_results_predicted.tsv'

ALL_TMs = []
ALL_SIM_VALUES = []
```

```python

search_obj = os.scandir(DIR_IN)

for dir_item in search_obj:
    if dir_item.is_file():
        
        FNAME_IN_ALL.append(DIR_IN + dir_item.name)

print(FNAME_IN_ALL)
```

```python
for fname in FNAME_IN_ALL:
    try:
        df = pd.read_csv (fname, sep = '\t')
        routes = read_reactions_dataframe(df, smiles_column = 'rxsmiles', group_by = ['route_index', 'tm_inchikey'], metadata_columns = ['mapped_reaction_smiles'])
        print (routes)
        # Rs = list(routes)
        # # print (Rs)
        # R = Rs[0]
        # print (R.atom_mapped_reaction_smiles())

        #print (routes.atom_mapped_reaction_smiles())
        # # #print (Rs)

    
        simmx = simple_route_similarity(routes)

        print (simmx)
        
        tm = list(df['tm_inchikey'])[0]

        
        for i in range(len(simmx)):
            for j in range(i + 1, len(simmx)):
                ALL_TMs.append(tm)
                ALL_SIM_VALUES.append(simmx[i][j])

    except:
        print (f'[W] No route could have been reconstructed in file: {fname} .')

df_res = pd.DataFrame({
                        'tm_inchikey': ALL_TMs,
                        'similarity': ALL_SIM_VALUES
                    })


print (df_res)

df_res.to_csv (FNAME_OUT, sep = '\t', index = False)

print ('[Done.]')

```
