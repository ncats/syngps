```python
# Env: syngps_rev
#import syngps
import json
#import networkx as nx
import pandas as pd

#import os
```

```python
FNAME_IN = '../data/input/case_study_3.json'
FNAME_OUT = '../data/output/cs3_smiles.tsv'
```

```python
unique_pairs = {}



with open(FNAME_IN) as f:
        routes_json = json.load(f)
        #print (f'{routes_json}')
        #print(json.dumps(routes_json, indent=4))

for n in routes_json['evidence_synth_graph']['nodes']:
        if n['node_type'] == 'substance':
                #print (n['inchikey'])
                #print (n['canonical_smiles'])

                unique_pairs[n['inchikey']] = n['canonical_smiles']

#print (unique_pairs)

inchikeys = []
smiles = []

for k, v in unique_pairs.items():
        inchikeys.append(k)
        smiles.append(v)



df = pd.DataFrame({
        'inchikey': inchikeys,
        'smiles': smiles
})


print (df)

df.to_csv(FNAME_OUT, sep = '\t', index = False)

print ('[Done.]')
```
