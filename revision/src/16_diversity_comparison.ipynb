```python
# Env: routesim
import pandas as pd
```

```python
FNAME_IN_EVIDENCE_LASM = '../data/output/similarity_results_evidence_lasm.tsv'

FNAME_IN_EVIDENCE_INV = '../data/output/similarity_results_evidence_inv.tsv'

FNAME_IN_PREDICTED = '../data/output/similarity_results_predicted.tsv'

FNAME_OUT = '../data/output/route_diversity_stats.tsv'
```

```python
def process_stats (fname, route_type):


    df = pd.read_csv(fname, sep = '\t')

    print (df.columns)

    df_stat1 = df.groupby(['tm_inchikey'], as_index = False)  \
                                    .agg ({
                                        'similarity': 'mean'
                                        })                                      \
                                    .rename(columns = {
                                        'similarity': 'mean_similarity'}
                                    )

    df_stat2 = df.groupby(['tm_inchikey'], as_index = False)  \
                                    .agg ({
                                        'similarity': 'std'
                                        })                                      \
                                    .rename(columns = {
                                        'similarity': 'std_similarity'}
                                    )                                           \
                                    .fillna (0)                                                                                                                                             

    df_stats = df_stat1.merge(df_stat2, on = 'tm_inchikey', how = 'inner')

    df_stat1 = df.groupby(['tm_inchikey'], as_index = False)  \
                                    .agg ('count')                              \
                                    .rename(columns = {
                                        'similarity': 'triangular_simmx_size'}
                                    )                                  

    df_stats = df_stats.merge(df_stat1, on = 'tm_inchikey', how = 'inner')


    df_stats['route_type'] = route_type

    #df_stats.to_csv(FNAME_OUT, sep = '\t', index = False)

    print (df_stats)

    return (df_stats)

```

```python
df_evidence_lasm_stats = process_stats (FNAME_IN_EVIDENCE_LASM, route_type = 'evidence_lasm')

df_evidence_inv_stats = process_stats (FNAME_IN_EVIDENCE_INV, route_type = 'evidence_inv')

df_predicted_stats = process_stats (FNAME_IN_PREDICTED, route_type = 'predicted')



print (f'Evidence - LASM similarity mean: {df_evidence_lasm_stats['mean_similarity'].mean()}, Evidence - Inv similarity mean: {df_evidence_inv_stats['mean_similarity'].mean()}, predicted similarity mean: {df_predicted_stats['mean_similarity'].mean()}')


df_stats = pd.concat([df_evidence_lasm_stats, df_evidence_inv_stats, df_predicted_stats])

df_stats.to_csv (FNAME_OUT, sep = '\t', index = False)
```
